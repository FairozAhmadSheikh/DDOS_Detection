# Lets first handle the Data 
`* Split data into features and targets `

`* Then into train test split `

```python
X=data.drop(columns=['Label', 'Class'])
y=data["Class"] # As we will be doing binary classfication for now and later we can go for multiclass 

from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)
print(X_train.shape , y_train.shape)
print(X_test.shape , y_test.shape)
```

# The dataset is imbalanced  `3.4 : 1`   percentage wise  77.3%  and 22.7 % 

build a model on this imabalanced dataset first and then we will fix it 

```python
from sklearn.svm import LinearSVC

clf=LinearSVC()   # Going for default settings first then we can improve it 

clf.fit(X_train,y_train)
```

```python
clf.score(X_test,y_test)
```

```python
y_preds=clf.predict(X_test)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf=RandomForestClassifier()

X=data.drop(columns=['Label', 'Class'])
y=data["Class"] # As we will be doing binary classfication for now and later we can go for multiclass 

from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)

rf.fit(X_train,y_train)
```

```python
rf.score(X_test,y_test)
```

```python
rf_preds=rf.predict(X_test)
```
