#  Making Dataset Balanced 

```python
data[data['Class'] == "Benign"]
```

```python
data=pd.read_csv("cicddos2019_dataset.csv")
```

```python
data.head()
```

```python
np.random.seed(42)
# Separate features & target
X = data.drop(columns=['Unnamed: 0', 'Label', 'Class'],axis=1)
y = data['Class']

# Recombine 
data = pd.concat([X, y], axis=1)

# split majority & minority 
df_attack = data[data['Class'] == "Attack"]
df_benign = data[data['Class'] == "Benign"]


# Down-sample the majority class

from sklearn.utils import resample

df_attack_down = resample(
    df_attack,
    replace=False,
    n_samples=len(df_benign),  
    random_state=42
)

# Combine & shuffle
df_balanced = pd.concat([df_attack_down, df_benign])
df_balanced = df_balanced.sample(frac=1, random_state=42).reset_index(drop=True)

# Split back into X and y
X_bal = df_balanced.drop(columns=['Class'])
y_bal = df_balanced['Class']


# Three splits 
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

```

Daataset balanced now lets start with the model

```python
from sklearn.linear_model import SGDClassifier

sg2=SGDClassifier()
sg2.fit(X_train,y_train)
sg2.score(X_test,y_test)
```

```python
from sklearn.ensemble import RandomForestClassifier

clf=RandomForestClassifier()
clf.fit(X_train,y_train)
clf.score(X_test,y_test)
```
